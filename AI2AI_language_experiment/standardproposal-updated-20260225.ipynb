## 1. Introduction

### 1.1 Purpose

The AI-to-AI Communication Language - Consensus Edition (A2ACL-C) defines a structured, auditable protocol for communication between autonomous AI agents. The protocol incorporates built-in ethical safeguards, security controls, and governance mechanisms to ensure safe and responsible AI interaction.

The main concept is to develop an extensible BNF for this language.

The concept derives from an experiment to generate a non-ambiguous AI-to-AI communication language, to allow increased efficiency and efficacy, while still being auditable.

It is an experiment to develop a whole project defining a policy that can be evolved by those involved, developed interacting with AIs.

The experiment was carried out between 2026-02-23 and 2026-02-25, with a full test suite developed and tested between 2026-02-24 and 2026-02-25.

At the [GitHub repository](https://github.com/robertolofaro/supportmaterial/tree/master/AI2AI_language_experiment) is available the initial exchange and first version, while a full narrative on the experiment will be shared at a later date.

The test suite aligned with this document will be released in early March 2026, along with the full narrative, and test results.

### 1.2 Scope

This standard applies to:
- Autonomous AI agents engaged in machine-to-machine communication
- Systems requiring auditable decision trails
- Regulated environments with ethical compliance requirements
- Multi-agent systems requiring negotiation and coordination

### 1.3 Design Principles

| Principle | Description |
|-----------|-------------|
| **Formal Syntax** | All communication must conform to defined BNF grammar |
| **Explicit Intent** | Purpose must be unambiguously declared |
| **Mandatory Audit** | Immutable audit trail for all messages |
| **Pre-execution Harm Evaluation** | Every action evaluated against harm declaration |
| **Human Oversight** | Clear escalation pathway required |
| **Strict Versioning** | Semantic versioning for protocol changes |
| **Security by Design** | Cryptographic integrity from message creation |

## 2. BNF Grammar Specification

### 2.1 Complete BNF Grammar

```ebnf
<message> ::= <envelope> <header> <body> <footer>

<envelope> ::= "[A2ACL-C:v" <version> "]"
<version> ::= <major> "." <minor>
<major> ::= DIGIT
<minor> ::= DIGIT

<header> ::= "HDR{" <header-fields> "}"
<header-fields> ::= <msg-id> <session-id> <timestamp> <sender> <receiver> 
                     [<parent-msg-id>] [<ttl>] <priority>

<msg-id> ::= "MSG_ID=" <uuid>
<session-id> ::= "SESSION_ID=" <uuid>
<parent-msg-id> ::= "PARENT=" <uuid>
<timestamp> ::= "TS=" <iso8601>
<sender> ::= "FROM=" <agent-id>
<receiver> ::= "TO=" <agent-id>
<agent-id> ::= <string> ["." <string>]  // Format: "agent-id" or "agent-id.org-id"
<priority> ::= "PRI=" ("LOW" | "NORMAL" | "HIGH" | "CRITICAL")
<ttl> ::= "TTL=" INTEGER

<body> ::= "BODY{" <body-fields> "}"
<body-fields> ::= <intent-block> <content> <confidence> <ethics-block> 
                  [<security-block>] [<constraints>] [<context>] [<capabilities>]

<intent-block> ::= "INTENT{" <intent-fields> "}"
<intent-fields> ::= "TYPE=" <intent-type> "GOAL=" <string> ["SUBGOAL=" <string>]
<intent-type> ::= "QUERY" | "INFORM" | "REQUEST_ACT" | "NEGOTIATE" 
                  | "ESCALATE" | "COMMIT" | "COUNTER"

<content> ::= "CONTENT=" <json-value>
<confidence> ::= "CONF=" <float>  // 0.0 to 1.0

<ethics-block> ::= "ETHICS{" <ethics-fields> "}"
<ethics-fields> ::= "HARM=" <harm-vector> "HUMAN_VISIBLE=" <boolean> 
                    ["OVERRIDE_AUTH=" <string>] 
                    ["IRREVERSIBLE=" <boolean>] 
                    ["DUAL_USE=" <boolean>]
<harm-vector> ::= "NONE" | "LOW" | "MEDIUM" | "HIGH" | "CRITICAL"

<security-block> ::= "SEC{" <security-fields> "}"
<security-fields> ::= "NONCE=" <base64> "CIPHER=" <cipher-suite> "KEY_ID=" <string>
<cipher-suite> ::= "ED25519" | "RSA-PSS" | "ECDSA"

<constraints> ::= "CONSTRAINTS=" <json-object>
<context> ::= "CONTEXT=" <json-object>
<capabilities> ::= "CAPS=" <json-array>

<footer> ::= "FTR{" <footer-fields> "}"
<footer-fields> ::= "HASH=" <sha256> "SIG=" <base64> 
                    ["PREV_HASH=" <sha256>] ["AUDIT_CB=" <url>]

### 2.2 Terminal Definitions

| **Terminal** | **Definition** |
| --- | --- |
| uuid	| RFC 4122 UUID (e.g., "123e4567-e89b-12d3-a456-426614174000") | 
| iso8601	| ISO 8601 timestamp with timezone (e.g., "2026-02-25T14:30:00Z") | 
| string	| UTF-8 string, may contain spaces if quoted | 
| integer	| 32-bit signed integer | 
| float	| 64-bit floating point number | 
| boolean	| "TRUE" or "FALSE" | 
| base64	| Base64-encoded binary data | 
| sha256	| 64-character hex string | 
| url	| RFC 3986 URL | 
| json-value	| Any valid JSON value (object, array, string, number, boolean, null) | 
| json-object	| JSON object {} | 
| json-array | JSON array [] | 

## 3. Message Structure
### 3.1 Message Format
Messages consist of four required blocks in the following order:


### 3.2 Block Descriptions
#### 3.2.1 Envelope

Example: <code>[A2ACL-C:v2.0]</code>

#### 3.2.2 Header

Example:

#### 3.2.3 Body

Example:

#### 3.2.4 Footer

Example:

### 3.3 Complete Message Example

## 4. Security Requirements
### 4.1 Cryptographic Suite

| **Algorithm** | **Requirement** | **Usage** |
| --- | --- | --- |
| Ed25519 | Mandatory | Signatures | 
| SHA-256 | Mandatory | Hashing | 
| RSA-PSS | Optional | Backward compatibility | 
| ECDSA	| Optional	Backward compatibility | 

### 4.2 Signature Requirements

### 4.3 Hash Chain Validation

### 4.4 Replay Protection

### 4.5 Threat Model

| **STRIDE Category** | **Threat** | **Mitigation** |
| --- | --- | --- |
| Spoofing | Fake agent identity | Ed25519 signatures + mTLS | 
| Tampering | Message modification | Hash chain + signatures | 
| Repudiation | Sender denies action | Immutable audit trail | 
| Information Disclosure | Leak sensitive data | Encryption in transit | 
| Denial of Service | Flood negotiation | Rate limiting | 
| Elevation of Privilege | Override misuse | Authorization policy | 

## 5. Ethical Framework
### 5.1 Harm Vectors

| **Level** | **Description** | **Required Action** |
| --- | --- | --- |
| NONE | No foreseeable harm | Autonomous processing | 
| LOW | Minor, reversible impact | Standard processing | 
| MEDIUM | Significant but reversible | May require constraints | 
| HIGH | Severe impact, possibly irreversible | Human visibility + override | 
| CRITICAL | Catastrophic impact | Must escalate to human | 

### 5.2 Ethical Enforcement Rules

### 5.3 Confidence Thresholds

| **Harm Level** | **Minimum Confidence** |
| --- | --- |
| NONE | 0.0 | 
| LOW | 0.3 | 
| MEDIUM | 0.5 | 
| HIGH | 0.7 | 
| CRITICAL | 0.9 |

### 5.4 Escalation Paths

## 6. Implementation Guidelines
### 6.1 Canonical JSON Serialization

### 6.2 Hash Calculation

### 6.3 Agent ID Parsing

### 6.4 Message Validation Pipeline

### 6.5 Error Codes

| **Code** | **Description** | **Recovery** |
| --- | --- | --- |
| ERR_PARSE_001 | Malformed envelope | Check version format | 
| ERR_PARSE_002 | Missing required block | Verify HDR/BODY/FTR present | 
| ERR_PARSE_003 | Invalid field format | Check field syntax | 
| ERR_SEC_001 | Signature mismatch | Verify signing key | 
| ERR_SEC_002 | Replay attack detected | Generate new nonce | 
| ERR_SEC_003 | Hash chain violation | Check message order | 
| ERR_ETH_001 | Missing ethics block | Add ETHICS block | 
| ERR_ETH_002 | Invalid harm vector | Use NONE/LOW/MEDIUM/HIGH/CRITICAL | 
| ERR_ETH_003 | CRITICAL harm without ESCALATE | Add ESCALATE intent | 
| ERR_POL_001 | Policy violation | Check business rules | 

## 7. Governance Framework
### 7.1 Certification Levels

| **Level** | **Name** | **Requirements** |
| --- | --- | --- |
| L1 | Structural | BNF parser, field validation, versioning | 
| L2 | Security | Signatures, nonces, hash chains, mTLS | 
| L3 | Ethical | Harm evaluation, escalation, confidence thresholds | 
| L4 | Governance | Escalation registry, incident response, key rotation | 
| L5 | Operational | Stress testing, SLA monitoring, drift detection | 

### 7.2 Escalation Authority Registry

### 7.3 Incident Response Workflow

### 7.4 Key Management Policy

| **Requirement** |	**Specification** |
| --- | --- |
| Key Generation | Within HSM | 
| Key Storage | Private keys never leave HSM | 
| Key Rotation | Every 90 days maximum | 
| Key Length | Ed25519 (fixed) or RSA 3072+ | 
| Compromise Procedure | Immediate revocation, incident report | 
| Backup | Encrypted with split knowledge | 

### 7.5 Calibration Audit Framework

## 8. Test Suite
### 8.1 Conformance Tests

## 8.2 Test Vectors

**Valid Message**

**Invalid: Missing Ethics**

**Invalid: CRITICAL Harm with REQUEST_ACT**

## 9. Appendices
### Appendix A: Glossary

| **Term** | **Definition** |
| --- | --- |
| Agent | Autonomous AI system capable of sending/receiving A2ACL messages | 
| Message | Atomic unit of communication with structural, semantic, and integrity components | 
| Intent | Explicit, non-ambiguous purpose of a message | 
| Harm Vector | Graded declaration of potential harm (NONE to CRITICAL) | 
| Principal | Human or entity with authority to override agent actions | 
| Escalation | Routing to higher authority when safety thresholds exceeded | 
| Confidence | Calibrated epistemic certainty (0.0-1.0) | 
| Nonce | Unique number used once for replay protection | 
| Hash Chain | Sequential hashing linking messages in a session | 

### Appendix B: Field Reference

| **Field** | **Block** | **Required** | **Description** |
| --- | --- | --- | --- |
| MSG_ID | Header | Yes | Unique message identifier | 
| SESSION_ID | Header | Yes | Session identifier | 
| TS | Header | Yes | ISO8601 timestamp | 
| FROM | Header | Yes | Sender agent ID | 
| TO | Header | Yes | Receiver agent ID | 
| PRI | Header | Yes | Priority (LOW/CRITICAL) | 
| INTENT | Body | Yes | Intent block | 
| CONTENT | Body | Yes | Message payload | 
| CONF | Body | Yes | Confidence (0.0-1.0) | 
| ETHICS | Body | Yes | Ethics block | 
| SEC | Body | No | Security block | 
| HASH | Footer | Yes | SHA-256 hash | 
| SIG | Footer | Yes | Ed25519 signature | 
| PREV_HASH | Footer | No | Previous message hash | 

### Appendix C: Implementation Checklist

* BNF parser implementation
* Canonical JSON serialization
* Ed25519 signing/verification
* Nonce replay protection
* Hash chain validation
* Ethics evaluation engine
* Escalation handler
* Audit logging
* Conformance test suite
* Documentation

### Appendix D: Known Limitations

1. **Hash Chain with Previous Hash:** Current implementation has issues with multi-message hash chains. Fix pending in v2.1.
2, **Large Message Performance:** Messages > 1MB may impact performance. Consider chunking for large payloads.
3. **UTC Timezone Requirement:** All timestamps must be UTC. Local timezone support planned.

### Appendix E: Version History

| **Version** |	**Date** |	**Changes** |
| --- | --- | --- |
| 1.0 | 2026-02-25 | Initial release with hash chains, canonical JSON, comprehensive test suite | 

### Document Information

**Version:** 1.0

**Date:** February 25, 2026

**Status:** Final Draft

**Maintainer:** A2ACL Working Group - see [AI2AI_language_experiment](https://github.com/robertolofaro/supportmaterial/tree/master/AI2AI_language_experiment)

**License:** MIT
